```python
from pathlib import Path
import time
import pandas as pd

ROOT = Path.cwd()  # folder containing run_pipeline.ipynb
rows = []
for path in (ROOT / 'd_phase1_train/output/phase1_runs').glob('*/history.csv'):
    history = pd.read_csv(path)
    modified = path.stat().st_mtime
    rows.append({'run': path.parent.name, 'saved_step': int(history['step'].iloc[-1]),
                 'val_loss': history['val_loss'].iloc[-1],
                 'last_saved': pd.Timestamp(modified, unit='s', tz='UTC').tz_convert('America/New_York'),
                 'minutes_since_save': round((time.time() - modified) / 60, 1)})

progress = pd.DataFrame(rows, columns=['run', 'saved_step', 'val_loss', 'last_saved', 'minutes_since_save'])
display(progress.sort_values('last_saved', ascending=False).head(40))
```
